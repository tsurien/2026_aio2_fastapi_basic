```python
print(f"--- URL GET 요청 시작: {target_url} ---")
try:
    # with 문을 사용하여 자동으로 리소스 정리
    with urllib.request.urlopen(target_url, timeout=10) as response:
        status = response.status
        print(f"상태 코드: {status}")

        headers = response.headers
        print(f"콘텐츠 타입: {headers.get('Content-Type', '알 수 없음')}")

        # 헤더에서 인코딩 정보 얻기 (없으면 utf-8 가정)
        encoding = headers.get_content_charset() or 'utf-8'
        print(f"감지된/가정된 인코딩: {encoding}")

        if 200 <= status < 300: # 성공적인 응답 확인
            html_bytes = response.read() # 본문을 바이트로 읽기

            # 바이트를 문자열로 디코딩
            html_content = html_bytes.decode(encoding)

            print("\n--- 응답 내용 (일부) ---")
            print(html_content[:500] + "...") # 앞 500자만 출력
        else:
            print(f"오류 응답 수신: 상태 코드 {status}")

except urllib.error.HTTPError as e:
    # HTTP 오류 처리 (예: 404, 500)
    print(f"HTTP 오류 발생: {e.code} {e.reason}")

except urllib.error.URLError as e:
    # URL 관련 오류 처리 (예: 호스트 찾을 수 없음, 네트워크 연결 불가)
    print(f"URL 오류 발생: {e.reason}")

except TimeoutError:
    print("요청 시간 초과")

except Exception as e:
    print(f"기타 오류 발생: {e}")
```

```python
import datetime
import urllib.parse
import urllib.request
import urllib.error

post_test_url = "http://httpbin.org/post"

# 전송할 데이터 (딕셔너리)
current_time_kst = datetime.datetime.now() # 현재 시간: 2025-04-22 09:45 KST

post_payload = {
    'username': 'urllib_tester',
    'location': 'Seoul', # 현재 위치 정보
    'timestamp': current_time_kst.isoformat()
}

# 데이터를 URL 인코딩하고 바이트로 변환
encoded_payload = urllib.parse.urlencode(post_payload).encode('utf-8')

print(f"\n--- URL POST 요청 시작: {post_test_url} ---")
print(f"전송할 데이터 (인코딩 전): {post_payload}")
print(f"전송할 데이터 (인코딩 후): {encoded_payload}")

try:
    with urllib.request.urlopen(
        post_test_url,
        data=encoded_payload,
        timeout=10
    ) as response:
        print(f"POST 응답 상태 코드: {response.status}")

        post_response_body = response.read().decode('utf-8')

        print("\n--- POST 응답 내용 (일부) ---")

        # httpbin.org/post는 전송된 데이터를 json 형태로 다시 보여줌
        print(post_response_body[:500] + "...")

except urllib.error.URLError as e:
    print(f"POST 요청 중 오류: {e.reason}")

except Exception as e:
    print(f"POST 중 기타 오류: {e}")
```
