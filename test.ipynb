```python
# 1. 일반 파이썬 클래스 (에러 안 남, 통제 불가능)
class NormalUser: # 객체(여기선 클래스) 정의
    name: str = "Alice"
user1 = NormalUser() # 객체 생성
user1.name = 5 # 객체생성 __init__(self, )
print("일반 파이썬 결과:", user1.name) # 유연성

# 근데 이렇게 안 쓸거임
```

```python
from pydantic import BaseModel

class PydanticUser(BaseModel): # 베이스모델에 있는 체크 기능을 가져다 쓴거임
    name: str

user2 = PydanticUser(name='5457')
user2.name, type(user2.name)
```

```python
# 1. 데이터 모델 정의
class UserRegister(BaseModel):
    user_id: int    # 필수 입력, 정수형
    user_name: str   # 필수 입력, 문자열       
    email: str      # 필수 입력, 문자열            
    is_admin: bool = False  # 선택 입력, 기본값은 False
```

```python
user = UserRegister(
    user_id=1,
    user_name="홍길동",
    email="hong@example.com"
)

user
```

```python
input_data = {
    "user_id": "1004",    # 문자열로 들어왔지만 int로 자동 변환(Parsing)됩니다.
    "user_name": "codemaster",
    "email": "code@example.com"
    # is_admin은 생략되었으므로 기본값 False가 적용됩니다.
}
```

```python
user1 = UserRegister(**input_data) # 언팩킹 (묶어서 넣어놓다)
user1.user_id, type(user1.user_id)
```

```python
user1.model_dump() # 딕셔너리 형태로 제공
```

```python
user1.model_dump_json() # 문자열 형태로 제공
```
